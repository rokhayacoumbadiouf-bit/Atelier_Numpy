```python
import numpy as np
```

1.CREATION DE LISTE PYTHON 

```python
Liste_temperature = [24.5 , 25.1 , 26.3 , 27.0 , 25.8 ]
```

2.CREATION DE TABLEAU NUMPY

```python
tab_temp=np.array(Liste_temperature)
```

3.MULTIPLICATION DES DEUX PAR "2"

```python
Liste_temperature= Liste_temperature*2
print(Liste_temperature)

tab_temp=tab_temp*2
print(tab_temp)
```

4.OBSERVATION

La différence observée se situe principalement au niveau des résultats. La multiplication par 2 double la liste et donne un tableau où chaque élément est multiplié par 2.

Partie 7 – Concaténation 
On suppose qu’on a un 2nd bâtiment avec 3 capteurs.  
1) Créer aléatoirement la matrice représentant les températures mesurées par les 3 capteurs 
durant une journée 
2) Déterminer de deux manières différentes la matrice qui fusionne les températures mesurées par 
les 8 capteurs durant une journée 

1.Créer aléatoirement la matrice représentant les températures mesurées par les 3 capteurs 
durant une journée

```python
temperatures_bat2 = np.random.randint(15, 45, (3, 24))

print(temperatures_bat2)
```

2.Déterminer de deux manières différentes la matrice qui fusionne les températures mesurées par 
les 8 capteurs durant une journée 

```python
temperatures_8 = np.concatenate((temperatures, temperatures_bat2), axis=0)

print(temperatures_8)
print(temperatures_8.shape)
```

```python
temperatures_8 = np.vstack((temperatures, temperatures_bat2))

print(temperatures_8)
print(temperatures_8.shape)
```
