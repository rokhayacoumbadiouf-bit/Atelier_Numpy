On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8 
1) Créer une liste Python stockant ces températures; 
2) Créer un tableau NumPy équivalent stockant ces températures ; 
3) Effectuer une multiplication par 2 sur ces deux collections; 
4) Quelle différence observe-t-on ? 

```python
%pip install numpy
```

```python
import numpy as np
```

1.CREATION DE LISTE PYTHON 

```python
Liste_temperature = [24.5 , 25.1 , 26.3 , 27.0 , 25.8 ]
```

2.CREATION DE TABLEAU NUMPY

```python
tab_temp=np.array(Liste_temperature)
```

3.MULTIPLICATION DES DEUX PAR "2"

```python
Liste_temperature= Liste_temperature*2
print(Liste_temperature)

tab_temp=tab_temp*2
print(tab_temp)
```

4.OBSERVATION

La différence observée se situe principalement au niveau des résultats. La multiplication par 2 double la liste et donne un tableau où chaque élément est multiplié par 2.

1.Afficher les températures supérieures à 30 °C 

```python
temp_chaud = temperatures[temperatures > 30]
print(temp_chaud)
```

2.Afficher les températures « normales » càd comprises entre 20 °C et 30 °C

```python
temp_normal = temperatures [(temperatures > 20) & (temperatures < 30)]
print (temp_normal)
```

3. Supposons qu'une température supérieure à 35 °C soit considérée comme anormale. 
Déterminer anomalies, un tableau de booléen précisant les anomalies 

```python
anomalies = temperatures > 35
print (anomalies)
print (temperatures[anomalies])
```

4.Déterminer le nombre de températures anormales à l’aide de la variable anomalies  

```python
anom_sum=sum(anomalies)
print("NB anomalies:",anom_sum)
```
