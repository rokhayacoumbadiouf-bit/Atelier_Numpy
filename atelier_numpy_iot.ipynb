Partie 1 : Listes Python vs tableaux NumPy 
On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8 
1) Créer une liste Python stockant ces températures; 
2) Créer un tableau NumPy équivalent stockant ces températures ; 
3) Effectuer une multiplication par 2 sur ces deux collections; 
4) Quelle différence observe-t-on ? 

```python
%pip install numpy
```

```python
import numpy as np
```

1.CREATION DE LISTE PYTHON 

```python
Liste_temperature = [24.5 , 25.1 , 26.3 , 27.0 , 25.8 ]
```

2.CREATION DE TABLEAU NUMPY

```python
tab_temp=np.array(Liste_temperature)
```

3.MULTIPLICATION DES DEUX PAR "2"

```python
Liste_temperature= Liste_temperature*2
print(Liste_temperature)

tab_temp=tab_temp*2
print(tab_temp)
```

4.OBSERVATION

La différence observée se situe principalement au niveau des résultats. La multiplication par 2 double la liste et donne un tableau où chaque élément est multiplié par 2.


Partie 2 – Création des données 
1) Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro. 
2) Créer un tableau contenant 10 valeurs égales à 1. 
3) Créer un tableau de 10 valeurs initialisées à 25. 
4) Créer une matrice identité de taille 5 × 5. 
5) Créer les 24 heures (de 0 à 23) de mesure d’une journée 
6) Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C  
7) Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles

1. Creer un tableau representant les mesures de température de 10 capteurs , initialement à zero


```python
tab_zero=np.zeros(10)
print(tab_zero)
```

2.Créer un tableau contenant 10 valeurs égales à 1. 

```python
tab_one=np.ones(10)
print(tab_one)
```

3.Créer un tableau de 10 valeurs initialisées à 25. 

```python
tab_full=np.full(10,25)
print(tab_full)
```

4.Créer une matrice identité de taille 5 × 5

```python
mat_identite=np.eye(5)
print(mat_identite)
```

5.Créer les 24 heures (de 0 à 23) de mesure d’une journée 

```python
tab_heures=np.arange(0,24)
print(tab_heures)
```

6.Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C

```python
tab_valeurs=np.linspace(20,30,10)
print(tab_valeurs)
```

7.Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles 

```python
np.random.seed(42)
tab_aleatoire=np.random.rand(100)
print(tab_aleatoire)
```
